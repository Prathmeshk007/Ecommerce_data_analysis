```python
import numpy as np
x.replace(["N/A","NULL",""],np.nan,inplace=True)
```

```python
#remove extra place
x=x.apply( lambda x: x.str.strip() if x.dtype=="object" else x)
```

```python
x["Customer_Name"]=x["Customer_Name"].str.title()
x["City"]=x["City"].str.title()
x["State"]=x["State"].str.title()
print(x["State"])
```

```python
#keep only valid email
x=x[x["Email"].str.contains("@",na=False)]
```

```python
x["Email"].loc[300:345]
```

```python
x["Order_Date"]=pd.to_datetime(x["Order_Date"],errors="coerce")
x["Delivery_Date"]=pd.to_datetime(x["Delivery_Date"],errors="coerce")
```

```python
x
```

```python
# convert into numeric column
c=["Qty","Unit_Price","Discount"]
for i in c:
    x[i]=pd.to_numeric(x[i])
```

```python
x["Phone"]=x["Phone"].astype(str)
x["Discount"]=x["Discount"].fillna(0,inplace=True)
x["Phone"]=x["Phone"].fillna("Unknown",inplace=True)
x["Delivery_Date"]=x["Delivery_Date"].fillna(x["Order_Date"])
```

```python
x
```

```python
x["sales"]=x["Qty"]*x["Unit_Price"]
```

```python
x
```

```python
x['Net_Amount']=x["sales"]-(x["sales"]*x["Discount"]/100)
```

```python
x["Profit"]=x["Net_Amount"]*20/100
```

```python
x
```

```python
x.to_excel("New_clean_ecommerce.xlsx", index=False)
```
